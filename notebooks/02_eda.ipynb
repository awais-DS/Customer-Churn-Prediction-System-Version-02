```python
import matplotlib.pyplot as plt

counts = df['Churn'].value_counts()

plt.figure(figsize=(6, 6))
plt.pie(
    counts, 
    labels=counts.index, 
    autopct='%1.1f%%', 
    startangle=90, 
    colors=['#66b3ff', '#ff9999'],
    explode=[0, 0.08]
)
plt.title('Customer Churn Distribution (%)', fontweight='bold')
plt.show()
```

```python
df["Churn"]=df["Churn"].map({"Yes": 1,"No":0})
```

## Bivariate Anlysis

```python
sns.barplot(data=df,x="Contract",y="Churn")
```

```python
plt.figure(figsize=(9, 5))
sns.barplot(data=df, x='Contract', y='Churn', hue='InternetService', errorbar=None)
plt.title('Churn Rate by Contract Type & Internet Service', fontweight='bold')
plt.ylabel('Churn Rate (Proportion)')
plt.show()
```

```python
plt.figure(figsize=(9, 5))
sns.scatterplot(data=df, x='tenure', y='MonthlyCharges', hue='Churn', alpha=0.6)
plt.title('Tenure vs. Monthly Charges by Churn Status', fontweight='bold')
plt.show()
```

```python
plt.figure(figsize=(6, 4))
cols = ['tenure', 'MonthlyCharges', 'Churn']
sns.heatmap(df[cols].corr(), annot=True, cmap='coolwarm', fmt='.2f', vmin=-1, vmax=1)
plt.title('Correlation Matrix', fontweight='bold')
plt.show()
```
